```python
from bs4 import BeautifulSoup
import requests
```

```python
name=[]
price=[]
desc=[]
reviewcount=[]
for i in range (1,5):
    page=requests.get(f'https://webscraper.io/test-sites/e-commerce/static/computers/tablets?page={i}')
    soup=BeautifulSoup(page.content,'html.parser')
    week=soup.find_all('div',class_='col-md-4 col-xl-4 col-lg-4')
    for k in week:
        n=k.find('a')
        name.append(n.get_text().strip())
        p=k.find('h4')
        price.append(p.get_text().strip())
        d=k.find('p',class_='description card-text')
        desc.append(d.get_text().strip())
        r=k.find(class_='review-count float-end')
        reviewcount.append(r.get_text().strip().split()[0])
    
   
```
